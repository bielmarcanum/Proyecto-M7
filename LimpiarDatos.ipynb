Se cargan Archivos con lo que se trabajaran


```python
#siguiendo la estructura del ejemplo de la practica de M3

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

ARCHIVO_ITER    = "conjunto_de_datos_iter_00CSV20.csv"
ARCHIVO_POBREZA = "Concentrado_indicadores_de_pobreza_2020.xlsx"
ARCHIVO_GRADES  = "Indicadores20261114093310.xls"

print("Archivos Cargados:")
print("1. ", ARCHIVO_ITER)
print("2. ", ARCHIVO_POBREZA)
print("3. ", ARCHIVO_GRADES)
```

Disponibilidad de los archivos

```python
archivos  = [ARCHIVO_ITER, ARCHIVO_POBREZA, ARCHIVO_GRADES]
disponible = True
for archivo in archivos:
  if Path(archivo).exists():
    print("Archivo encontrado",archivo)
  else:
    print("Anchivo NO encontrado", archivo)
    disponible  = False
if disponible:
  print("\nArchivos Disponibles")
else:
  print("\nError: Archivo faltante")
```

VERIFICANDO COLUMNAS DE ITER


```python
vista_ITER = pd.read_csv(ARCHIVO_ITER, nrows=5, low_memory=False)

print("Número total de columnas del archivo:", len(vista_ITER.columns))
print("\nPrimeros nombres de columnas:")
print(vista_ITER.columns[:250].tolist())
```

VERIFICANDO COLUMNAS DE POBREZA

```python
vista_POBREZA = pd.read_excel(ARCHIVO_POBREZA, header=[4, 5], nrows=10)

# 2. Limpiamos y unimos los dos niveles de encabezado
nuevas_columnas = []
for nivel_superior, nivel_inferior in vista_POBREZA.columns:
    # Si el nivel inferior dice 'Unnamed' (pasa en las columnas B a H combinadas), nos quedamos solo con el de arriba
    if 'Unnamed' in str(nivel_inferior):
        nuevas_columnas.append(str(nivel_superior).strip())
    # Si el nivel superior dice 'Unnamed', nos quedamos con el de abajo
    elif 'Unnamed' in str(nivel_superior):
        nuevas_columnas.append(str(nivel_inferior).strip())
    # Si ambos tienen texto (de la columna I en adelante), los unimos con un guion bajo
    else:
        nuevas_columnas.append(f"{str(nivel_superior).strip()}_{str(nivel_inferior).strip()}")

vista_POBREZA.columns = nuevas_columnas

print("Número total de columnas del archivo:", len(vista_POBREZA.columns))
print("\nPrimeros nombres de columnas:\n")
print(vista_POBREZA.columns.tolist())
```

VERIFICANDO COLUMNAS DEL PROMEDIO DE GRADO ESCOLAR
SE PUEDE EXTRAER DE ITER NO HACE FALTA ESTE ARCHIVO
PROCEDEMOS A EXTRAER COLUMNAS


```python
COLUMNAS_ITER = [
    "ENTIDAD", "NOM_ENT",
    "MUN", "NOM_MUN", "LOC",
    "POBTOT", "GRAPROES",
    "VPH_PC", "VPH_CEL",
    "VPH_INTER",
    "VPH_SINCINT", "VIVPARH_CV"
]
COLUMNAS_POBREZA = [
    "Municipio","Clave de entidad",
    "Clave de municipio",
    "Pobreza_Porcentaje\n2020"
]

# 1. Cargamos ITER de forma normal
ITER = pd.read_csv(ARCHIVO_ITER, usecols=COLUMNAS_ITER, low_memory=False)

# 2. Cargamos POBREZA con encabezado de dos niveles para luego limpiarlo y filtrar las columnas
_pobreza_raw = pd.read_excel(ARCHIVO_POBREZA, header=[4, 5])

nuevas_columnas = []
for nivel_superior, nivel_inferior in _pobreza_raw.columns:
    if 'Unnamed' in str(nivel_inferior):
        nuevas_columnas.append(str(nivel_superior).strip())
    elif 'Unnamed' in str(nivel_superior):
        nuevas_columnas.append(str(nivel_inferior).strip())
    else:
        nuevas_columnas.append(f"{str(nivel_superior).strip()}_{str(nivel_inferior).strip()}")

_pobreza_raw.columns = nuevas_columnas

# Ahora que las columnas están aplanadas y normalizadas, podemos filtrar las que queremos
POBREZA = _pobreza_raw[COLUMNAS_POBREZA]

print("Columnas extraidas de ITER:")
print(ITER.columns.tolist())
print("\nColumnas extraidas de POBREZA:")
print(POBREZA.columns.tolist())

#Mostrar los encabezados

print("\nFilas ITER:")
display(ITER.head())
print("\nFilas POBREZA:")
display(POBREZA.head())
```

LIMPIAR FILAS


```python
for columna in ["ENTIDAD", "MUN", "LOC"]:
  ITER[columna] = pd.to_numeric(ITER[columna], errors='coerce')

# Filtramos las filas municipales de ITER (LOC == 0)
ITER_municipal = ITER[
    (ITER["ENTIDAD"] > 0) &
    (ITER["MUN"] > 0) &
    (ITER["LOC"] == 0)
].copy()

POBREZA = POBREZA.copy()

for columna in ["Clave de municipio", "Clave de entidad"]:
  POBREZA[columna] = pd.to_numeric(POBREZA[columna], errors='coerce')

# Filtramos las filas de POBREZA de forma segura (las claves municipales son mayores a 1000)
POBREZA_municipal = POBREZA[
    (POBREZA["Clave de municipio"] > 1000) &
    (POBREZA["Clave de municipio"].notna()) &
    (POBREZA["Clave de entidad"]>0)
].copy()

print("Filas municipales obtenidas (ITER):", len(ITER_municipal))
display(ITER_municipal.head())

print("\nFilas municipales obtenidas (POBREZA):", len(POBREZA_municipal))
display(POBREZA_municipal.head())
```

Creando llave CVEGEO

```python
ITER_municipal["CVE_ENT"] = (
    ITER_municipal["ENTIDAD"].astype("Int64").astype(str).str.zfill(2)
)

ITER_municipal["CVE_MUN"] = (
    ITER_municipal["MUN"].astype("Int64").astype(str).str.zfill(3)
)

ITER_municipal["CVEGEO"] = (
    ITER_municipal["CVE_ENT"] + ITER_municipal["CVE_MUN"]
)

display(
    ITER_municipal[
        ["CVEGEO", "NOM_ENT", "NOM_MUN"]
    ].head(10)
)
```

Construimos el porcentaje para cada variable
pkm =100*N_km/Dm

```python
for columna in ["VIVPARH_CV", "VPH_INTER", "VPH_PC"]:
    ITER_municipal[columna] = pd.to_numeric(
        ITER_municipal[columna].replace({"*": np.nan}),
        errors="coerce"
    )
#Porcentaje de viviendas con internet; componente principal
ITER_municipal["P_Im"] = (
    ITER_municipal["VPH_INTER"] / ITER_municipal["VIVPARH_CV"] * 100
)

#Porcentaje de viviendas con computadora, laptop o tableta; componente principal
ITER_municipal["P_Em"] = (
    ITER_municipal["VPH_PC"] / ITER_municipal["VIVPARH_CV"] * 100
)


base_censo = ITER_municipal[
    [
        "CVEGEO", "CVE_ENT", "CVE_MUN",
        "NOM_ENT", "NOM_MUN",
        "POBTOT",
        "VIVPARH_CV", "VPH_INTER",
        "P_Im", "P_Em", "GRAPROES"
    ]
].copy()

base_censo = base_censo.rename(columns={
    "NOM_ENT": "ENTIDAD",
    "NOM_MUN": "MUNICIPIO"
})

display(base_censo.head())
```

Crear llave CVEGEO para pobreza


```python

#Municipio 	Clave de municipio 	Pobreza_Porcentaje\n2020
# En el concentrado de indicadores de pobreza ya esta construido el CVEGEO por lo tanto:

POBREZA["CVEGEO"] = (
    pd.to_numeric(POBREZA["Clave de municipio"], errors="coerce")
    .astype("Int64")
    .astype(str)
    .str.zfill(5)
)

display(
    POBREZA[
        ["CVEGEO", "Municipio", "Clave de entidad", "Clave de municipio", "Pobreza_Porcentaje\n2020"]
    ].head()
)

```

UNIR FUENTES

```python
DATABASE = base_censo.merge(
    POBREZA[["CVEGEO", "Pobreza_Porcentaje\n2020"]],
    on="CVEGEO",
    how="left"
)

display(DATABASE.head())
```

SE CONSTRUYE EL INDICE DE ACCESO DIGITAL

\begin{equation}
IAD_m=\frac{P_{Im}+P_{Em}}{2}
\end{equation}


```python
DATABASE["IAD"] =  (DATABASE["P_Im"] + DATABASE["P_Em"]) /2

display(
    DATABASE[
        [
            "CVEGEO", "ENTIDAD", "MUNICIPIO",
            "P_Im", "P_Em", "IAD"
        ]
    ].head(10)
)
```

CONTROL DE CALIDAD

```python
control = pd.DataFrame({
    "Pregunta": [
        "¿Cuántas filas tiene la base?",
        "¿Cuántas CVEGEO están duplicadas?",
        "¿Cuántos Pobreza_Porcentaje\n2020 están vacíos?",
        "¿Cuántos Pobreza_Porcentaje\n2020 =n.d.?",
        "¿Cuántos GRAPROES están vacíos?",
        "¿Cuántos municipios tienen VIVPARH_CV <= 0?"
    ],
    "Resultado": [
        len(DATABASE),
        int(DATABASE["CVEGEO"].duplicated().sum()),
        int(DATABASE["Pobreza_Porcentaje\n2020"].isna().sum()),
        str((DATABASE["Pobreza_Porcentaje\n2020"] == "n.d.").sum()),
        int(DATABASE["GRAPROES"].isna().sum()),
        int((DATABASE["VIVPARH_CV"] <= 0).sum())
    ]
})

display(control)

```

PREPARANDO BASE DE DATOS

```python
COLUMNAS_SALIDA = [
    "CVEGEO", "CVE_ENT", "CVE_MUN",
    "ENTIDAD", "MUNICIPIO",
    "POBTOT",
    "VPH_INTER", "VIVPARH_CV",
    "P_Im", "P_Em", "IAD", "GRAPROES", "Pobreza_Porcentaje\n2020"
]

DATABASE_final = DATABASE[COLUMNAS_SALIDA].copy()

print("BASE DE DATOS FINAL")
print("Filas:", DATABASE_final.shape[0])
print("Columnas:", DATABASE_final.shape[1])

display(DATABASE_final.head(10))
```

DESCARGA EN .XLSX


```python
diccionario = pd.DataFrame([
    ["CVEGEO", "Clave geográfica municipal de 5 dígitos", "Identificador"],
    ["CVE_ENT", "Clave de entidad federativa", "Identificador"],
    ["CVE_MUN", "Clave municipal", "Identificador"],
    ["ENTIDAD", "Nombre de la entidad federativa", "Texto"],
    ["MUNICIPIO", "Nombre del municipio", "Texto"],
    ["POBTOT", "Población total municipal", "Personas"],
    ["VPH_INTER", "Viviendas particulares habitadas que disponen de internet", "Personas"],
    ["VIVPARH_CV", "Total de viviendas particulares habitadas con caracteristicas", "Personas"],
    ["P_Im", "Porcentaje de viviendas que disponen de internet", "Porcentaje"],
    ["P_Em", "Porcentaje de viviendas que disponen de equipo de computo", "Porcentaje"],
    ["IAD", "Índice de Acceso Digital", "Índice"],
    ["GRAPROES", "Grado promedio de escolaridad", "Categoría"],
    ["Pobreza_Porcentaje\n2020", "Porcentaje de Pobreza municipal en 2020", "Porcentaje"]
], columns=["Variable", "Significado", "Tipo"])

NOMBRE_SALIDA = "Base_Final_M7.xlsx"

with pd.ExcelWriter(NOMBRE_SALIDA, engine="openpyxl") as writer:
    DATABASE_final.to_excel(writer, sheet_name="Base_Municipal", index=False)
    control.to_excel(writer, sheet_name="Control_Calidad", index=False)
    diccionario.to_excel(writer, sheet_name="Diccionario", index=False)

print("✓ Archivo creado:", NOMBRE_SALIDA)

try:
    from google.colab import files
    files.download(NOMBRE_SALIDA)
except ImportError:
    print("Archivo guardado en:", Path(NOMBRE_SALIDA).resolve())
```
