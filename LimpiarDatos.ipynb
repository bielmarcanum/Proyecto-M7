Se cargan Archivos con lo que se trabajaran


```python
#siguiendo la estructura del ejemplo de la practica de M3

import pandas as pd
import numpy as np
from pathlib import Path
from IPython.display import display

ARCHIVO_ITER    = "conjunto_de_datos_iter_00CSV20.csv"
ARCHIVO_POBREZA = "Concentrado_indicadores_de_pobreza_2020.xlsx"
ARCHIVO_GRADES  = "Indicadores20261114093310.xls"

print("Archivos Cargados:")
print("1. ", ARCHIVO_ITER)
print("2. ", ARCHIVO_POBREZA)
print("3. ", ARCHIVO_GRADES)
```

Disponibilidad de los archivos

```python
archivos  = [ARCHIVO_ITER, ARCHIVO_POBREZA, ARCHIVO_GRADES]
disponible = True
for archivo in archivos:
  if Path(archivo).exists():
    print("Archivo encontrado",archivo)
  else:
    print("Anchivo NO encontrado", archivo)
    disponible  = False
if disponible:
  print("\nArchivos Disponibles")
else:
  print("\nError: Archivo faltante")
```

VERIFICANDO COLUMNAS DE ITER


```python
vista_ITER = pd.read_csv(ARCHIVO_ITER, nrows=5, low_memory=False)

print("Número total de columnas del archivo:", len(vista_ITER.columns))
print("\nPrimeros nombres de columnas:")
print(vista_ITER.columns[:250].tolist())
```

VERIFICANDO COLUMNAS DE POBREZA

```python
vista_POBREZA = pd.read_excel(ARCHIVO_POBREZA, header=[4, 5], nrows=10)

# 2. Limpiamos y unimos los dos niveles de encabezado
nuevas_columnas = []
for nivel_superior, nivel_inferior in vista_POBREZA.columns:
    # Si el nivel inferior dice 'Unnamed' (pasa en las columnas B a H combinadas), nos quedamos solo con el de arriba
    if 'Unnamed' in str(nivel_inferior):
        nuevas_columnas.append(str(nivel_superior).strip())
    # Si el nivel superior dice 'Unnamed', nos quedamos con el de abajo
    elif 'Unnamed' in str(nivel_superior):
        nuevas_columnas.append(str(nivel_inferior).strip())
    # Si ambos tienen texto (de la columna I en adelante), los unimos con un guion bajo
    else:
        nuevas_columnas.append(f"{str(nivel_superior).strip()}_{str(nivel_inferior).strip()}")

vista_POBREZA.columns = nuevas_columnas

print("Número total de columnas del archivo:", len(vista_POBREZA.columns))
print("\nPrimeros nombres de columnas:\n")
print(vista_POBREZA.columns.tolist())
```

VERIFICANDO COLUMNAS DEL PROMEDIO DE GRADO ESCOLAR
SE PUEDE EXTRAER DE ITER NO HACE FALTA ESTE ARCHIVO
PROCEDEMOS A EXTRAER COLUMNAS


```python
COLUMNAS_ITER = [
    "ENTIDAD", "NOM_ENT",
    "MUN", "NOM_MUN", "LOC",
    "POBTOT", "GRAPROES",
    "VPH_PC", "VPH_CEL",
    "VPH_SINCINT"
]
COLUMNAS_POBREZA = [
    "Municipio",
    "Clave de municipio",
    "Pobreza_Porcentaje\n2020"
]

# 1. Cargamos ITER de forma normal
ITER = pd.read_csv(ARCHIVO_ITER, usecols=COLUMNAS_ITER, low_memory=False)

# 2. Cargamos POBREZA con encabezado de dos niveles para luego limpiarlo y filtrar las columnas
_pobreza_raw = pd.read_excel(ARCHIVO_POBREZA, header=[4, 5])

nuevas_columnas = []
for nivel_superior, nivel_inferior in _pobreza_raw.columns:
    if 'Unnamed' in str(nivel_inferior):
        nuevas_columnas.append(str(nivel_superior).strip())
    elif 'Unnamed' in str(nivel_superior):
        nuevas_columnas.append(str(nivel_inferior).strip())
    else:
        nuevas_columnas.append(f"{str(nivel_superior).strip()}_{str(nivel_inferior).strip()}")

_pobreza_raw.columns = nuevas_columnas

# Ahora que las columnas están aplanadas y normalizadas, podemos filtrar las que queremos
POBREZA = _pobreza_raw[COLUMNAS_POBREZA]

print("Columnas extraidas de ITER:")
print(ITER.columns.tolist())
print("\nColumnas extraidas de POBREZA:")
print(POBREZA.columns.tolist())

#Mostrar los encabezados

print("\nFilas ITER:")
display(ITER.head())
print("\nFilas POBREZA:")
display(POBREZA.head())
```

LIMPIAR FILAS


```python
for columna in ["ENTIDAD", "MUN", "LOC"]:
  ITER[columna] = pd.to_numeric(ITER[columna], errors='coerce')

# Filtramos las filas municipales de ITER (LOC == 0)
ITER_municipal = ITER[
    (ITER["ENTIDAD"] > 0) &
    (ITER["MUN"] > 0) &
    (ITER["LOC"] == 0)
].copy()

POBREZA = POBREZA.copy()

for columna in ["Clave de municipio"]:
  POBREZA[columna] = pd.to_numeric(POBREZA[columna], errors='coerce')

# Filtramos las filas de POBREZA de forma segura
POBREZA_municipal = POBREZA[
    (POBREZA["Clave de municipio"] > 1000) &
    (POBREZA["Clave de municipio"].notna())
].copy()

print("Filas municipales obtenidas (ITER):", len(ITER_municipal))
display(ITER_municipal.head())

print("\nFilas municipales obtenidas (POBREZA):", len(POBREZA_municipal))
display(POBREZA_municipal.head())
```
